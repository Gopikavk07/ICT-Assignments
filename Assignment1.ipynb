```python
temperatures = [30, 32, 29, 28, 31, 27, 26]
humidity = [60, 65, 58, 70, 75, 80, 68]


def average_temperature():
    return sum(temperatures) / len(temperatures)

def average_humidity():
    return sum(humidity) / len(humidity)

def weather_condition(temp, hum):
    if hum > 70:
        print("Rainy")
    elif temp > 32:
        print("Hot & Sunny")
    elif temp < 25:
        print("Cold")
    else:
        print("Sunny")

def analyze_weather():
    print("---- WEATHER DATA ANALYSIS ----")

    avg_temp = average_temperature()
    avg_hum = average_humidity()
    condition = weather_condition(avg_temp, avg_hum)

    print(f"Average Temperature: {avg_temp:.2f}°C")
    print(f"Average Humidity: {avg_hum:.2f}%")
    print(f"Overall Weather Condition: {condition}")

analyze_weather()
```
